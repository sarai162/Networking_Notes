## 1. IOS Command Modes

| Mode | Prompt | Purpose | Enter with |
| --- | --- | --- | --- |
| User EXEC | `>` | Limited read-only, basic monitoring | default on login |
| Privileged EXEC | `#` | Full read access | `enable` / `en` |
| Global Config | `(config)#` | Modify device configuration | `configure terminal` |
| Interface Config | `(config-if)#` | Configure one interface | `interface g0/1` |

* Progression: User EXEC → Privileged EXEC → Global Config → Specific Config
* Always check the prompt — know which mode you're in

## 2. Essential Commands

**Navigation**

| Command | Purpose |
| --- | --- |
| `enable` | Enter privileged mode |
| `configure terminal` | Enter config mode |
| `exit` | Go back one level |
| `end` | Return to privileged mode |
| `do <command>` | Run EXEC command from config mode |

**View**

| Command | Purpose |
| --- | --- |
| `show running-config` | Active config (RAM) |
| `show startup-config` | Saved config (NVRAM) |
| `show vlan brief` | VLANs and port assignments |

**Save**

| Command | Purpose |
| --- | --- |
| `write memory` | Save running-config → startup-config |
| `wr` | Shorthand for `write memory` |

**VLAN**

| Command | Purpose |
| --- | --- |
| `vlan 10` | Create VLAN 10 |
| `no vlan 10` | Delete VLAN 10 |
| `show vlan brief` | Verify VLAN configuration |

**Interface / Device**

| Command | Purpose |
| --- | --- |
| `interface g0/1` | Enter interface config mode |
| `shutdown` / `no shutdown` | Disable / enable interface |
| `hostname NAME` | Set device hostname |

## 3. Shortcuts & Productivity

* Abbreviations: `en` = enable, `conf t` = configure terminal, `int` = interface, `wr` = write memory
* TAB → auto-complete commands
* `?` → context-sensitive help
* Abbreviation must be unambiguous — if it matches multiple commands, IOS rejects it

## 4. IOS Error Messages

| Error | Meaning | Fix |
| --- | --- | --- |
| Invalid input | Command not recognized; `^` marks the error location | Check spelling / mode |
| Incomplete command | Command recognized, parameters missing | Add required information |
| Ambiguous command | Abbreviation matches multiple commands | Use more characters |

## 5. Port States

| State | Meaning |
| --- | --- |
| connected | Port active and working |
| notconnect | Port enabled, no device connected |
| administratively down | Port disabled by admin (`shutdown`) |

## 6. VLANs (Virtual LANs)

* Purpose: logically separate devices on the same physical switch
* Default: VLAN 1 exists on every switch
* Key concept: VLANs create broadcast domains without physical separation
* Why: security, broadcast control, organizational separation
* Advanced (senior interviews): trunk vs access ports, 802.1Q tagging, inter-VLAN routing

## 7. Running-Config vs Startup-Config

| | Running-Config | Startup-Config |
| --- | --- | --- |
| Location | RAM | NVRAM |
| Effect | Active, immediate | Loaded on device boot |
| Survives reboot | No (unless saved) | Yes |

* Critical rule: always save with `write memory` after changes

## 8. The "no" Command (undo pattern)

* Golden rule: prefix any command with `no` to remove/undo it
* `no vlan 10` → delete VLAN 10
* `no shutdown` → enable interface
* `no hostname` → remove hostname
* `no ip address` → remove IP configuration

## 9. Best Practices Checklist

* Check the prompt → know which mode you're in
* Read before configuring → `show` commands first
* Verify after configuring → confirm changes took effect
* Save your work → `wr`, or changes vanish on reboot
* TAB completion → avoid typos
* ↑ command history → efficiency
* Test abbreviations → must be unique
* Document changes
* Know running vs startup → when changes persist
* Use `no` for rollback

## 10. Interview Prep Map

**Must-know concepts**

* IOS mode hierarchy — when to use each, privilege levels, security implications
* Config persistence — running vs startup, why changes are lost without saving
* VLAN fundamentals — segmentation, security, broadcast control, VLAN 1 default, trunk vs access
* Troubleshooting approach — verify state first (`show`), interpret errors, rollback with `no`
* Interface management — shutdown vs no shutdown, port states, VLAN assignment
* Efficiency — shortcuts, TAB, `do` command
* Senior roles — config backup/restore/versioning, privileged-mode passwords, port security, 802.1Q, inter-VLAN routing

**Red flags to avoid**

* Confusing running-config with startup-config
* Not knowing how to save configurations
* Unable to explain VLAN purpose
* Not understanding the `no` command pattern
* Not verifying after configuration changes

**Scenario drills**

* Port shows "administratively down" → admin-disabled; fix with `no shutdown`
* Changes disappeared after reboot → never saved; running-config only lives in RAM
* Quickly verify all VLANs on a switch → `show vlan brief`
* Disable all ports in VLAN 10 → shut each member interface (or delete the VLAN — know the difference)

## 11. Quick-Win Answers (say-aloud)

* **First thing before configuring a switch?** → "Run `show running-config` and `show vlan brief` to understand the current state."
* **How do changes survive a reboot?** → "`write memory` or `copy running-config startup-config` — saves to NVRAM."
* **What does the `^` symbol mean in IOS?** → "It marks where the IOS parser detected the error in the command."
* **Why use VLANs?** → "To logically segment networks for security, broadcast control, and organizational separation — without physical infrastructure changes."
