## 1. What is a Network?

* Collection of devices connected to exchange data — laptop, phone, TV, printer, server
* If devices can exchange data, they form a network
* Entire purpose of networking: **send data + receive data** — everything exists for this

**By scale**

* LAN — one building/office (local network)
* WAN — between cities/countries (the internet is the biggest WAN)

**The shape of a home network**

`Laptop / Phone / TV → Switch → Router → Internet`

## 2. Protocols — The Rules of Communication

* Humans use languages → devices need rules → rules are called **protocols**
* Protocol = set of rules for communication; without them devices cannot understand each other

| Protocol | Purpose |
| --- | --- |
| HTTP | websites |
| HTTPS | secure (encrypted) HTTP |
| DNS | names → IP addresses |
| DHCP | auto-assign IP addresses |
| SMTP | sending email |
| FTP | file transfer |
| TCP | reliable delivery |
| UDP | fast delivery |

## 3. NIC & MAC Address — Device Identity

* NIC (Network Interface Card) = hardware that lets a device join a network — WiFi card, Ethernet adapter, mobile adapter
* Every NIC has a **MAC address** — e.g. `AA:BB:CC:11:22:33`

**Think:** MAC = Aadhaar number — unique, hardware-based identity

* Answers **"Who are you?"** — identity, NOT location
* Unique, usually permanent, used **inside the local network** (by switches)

## 4. IP Address — Device Location

* Identifies **where** a device sits on a network — e.g. `192.168.1.10`
* Answers **"Where are you?"** — location, NOT identity
* Used by routers, for communication **between networks**
* Can change (device moves, DHCP reassigns)

**Interview extras**

* Private ranges (LAN only, not routable on internet): `10.x.x.x`, `172.16–31.x.x`, `192.168.x.x`
* IPv4 — 32-bit, dotted decimal (running out) vs IPv6 — 128-bit, colon format (the fix)

## 5. MAC vs IP (classic interview confusion)

| | MAC | IP |
| --- | --- | --- |
| Answers | Who are you? | Where are you? |
| Level | hardware (Layer 2) | logical (Layer 3) |
| Format | `AA:BB:CC:11:22:33` | `192.168.1.10` |
| Changes? | permanent (usually) | can change |
| Used by | switches | routers |
| Scope | local network | across networks |

**Memory hook:** MAC = name on your ID card, IP = your current home address

## 6. Switch — Local Traffic

* Connects devices **within one network** (LAN)
* Learns **MAC → port** mappings in its MAC address table: `AA:BB → port 1`, `CC:DD → port 2`
* On arrival: checks the MAC table → forwards only to the correct port
* Benefit: efficient — no traffic flooding everywhere; works at **Layer 2**

**Think:** switch = local postman who knows every house in the neighborhood

**Hub vs Switch (interview add)**

* Hub — dumb: broadcasts every bit to ALL ports, everyone shares one collision domain
* Switch — smart: forwards only to the target port, per-port collision domains

## 7. Router — Between Networks

* Connects **different networks** — office LAN ↔ internet, branch ↔ HQ, LAN ↔ cloud
* Without a router: local communication works, internet impossible
* Jobs: path selection, forwarding traffic; works at **Layer 3**

**Think:** switch = inside one neighborhood, router = between cities

**Switch vs Router (confusion table)**

| | Switch | Router |
| --- | --- | --- |
| Connects | devices in the same network | different networks |
| Uses | MAC addresses | IP addresses |
| Layer | 2 | 3 |
| Analogy | neighborhood postman | inter-city courier hub |

## 8. DNS — The Internet Phonebook

* Humans remember names, computers use IPs → **DNS converts name → IP**
* Example: `google.com` → `142.250.x.x`

**Think:** DNS = phonebook — you know the name, DNS gives the number

**Interview extras**

* Record types: A (IPv4), AAAA (IPv6), CNAME (alias), MX (mail), NS (name servers)
* Symptom of DNS failure: internet works by IP address but not by name
* DNS itself uses port 53

## 9. DHCP — Automatic IP Assignment

* Without DHCP: manually set IP, subnet mask, gateway, DNS on **every** device
* With DHCP: router/server hands them out automatically — easy, fast, fewer errors

**DORA process (interview favorite)**

1. **D**iscover — client broadcasts: "any DHCP server out there?"
2. **O**ffer — server offers an IP
3. **R**equest — client accepts the offer
4. **A**cknowledge — server confirms the lease

## 10. What Happens When You Open YouTube? (THE interview question)

1. Browser asks the DNS server → DNS returns the IP (e.g. `142.x.x.x`)
2. Browser opens an HTTPS connection to that IP
3. Request travels: `Laptop → Switch → Router → Internet → YouTube server`
4. Server returns videos, images, text, data
5. Response returns through the same path in reverse; browser renders the page

**Say-aloud version:** "DNS lookup first, then HTTPS to the server's IP; the request travels switch → router → internet; the response returns the same path; the browser renders."

* Every topic in these notes is one step in this journey — this is the story to connect the dots

## 11. OSI & TCP/IP Models (added — must-know)

* Models split "how data travels" into layers — each layer has one job and talks to the layers above/below
* **OSI (7 layers)**, top → down: Application, Presentation, Session, Transport, Network, Data Link, Physical
* Mnemonic (top→down): **A**ll **P**eople **S**eem **T**o **N**eed **D**ata **P**rocessing
* **TCP/IP (4 layers)** — the practical model actually used: Application | Transport | Internet | Network Access

**How they map**

| OSI | TCP/IP |
| --- | --- |
| 5–7 Application/Presentation/Session | Application |
| 4 Transport | Transport |
| 3 Network | Internet |
| 1–2 Data Link + Physical | Network Access |

**Where things live (frequent interview Q)**

* Switch → Layer 2 | Router → Layer 3
* MAC → Layer 2 | IP → Layer 3 | TCP/UDP → Layer 4 | HTTP(S)/DNS → Layer 7
* **Frame** = Layer 2 data unit | **Packet** = Layer 3 data unit

* Interview trap: OSI is the teaching/reference model — nobody runs it directly; TCP/IP is what the internet uses

## 12. TCP vs UDP + Ports (added — must-know)

| | TCP | UDP |
| --- | --- | --- |
| Delivery | reliable — guaranteed, ordered | best-effort — no guarantee |
| Method | handshake (SYN → SYN-ACK → ACK) + retransmissions | fire-and-forget |
| Speed | slower | faster |
| Used for | HTTP/HTTPS, email, file transfer | video, gaming, voice, DNS |

**Think:** TCP = registered post with delivery confirmation; UDP = shouting and moving on

**Port = which application** gets the data on a device (IP finds the device, port finds the app)

**Common ports**

| Port | Service |
| --- | --- |
| 80 | HTTP |
| 443 | HTTPS |
| 53 | DNS |
| 25 | SMTP |
| 22 | SSH |
| 3389 | RDP |
| 20/21 | FTP |

## 13. Addressing Extras (added)

* **Private IP** — inside a LAN (`192.168.x.x` etc.), not routable on the internet | **Public IP** — globally unique, routable
* **Subnet mask** — splits an IP into network part + host part (`/24` = `255.255.255.0`)
* **Subnetting** — dividing one network into smaller networks — better security, less broadcast waste
* **Default gateway** — the router's IP; where your device sends any traffic destined for OTHER networks

**NAT (Network Address Translation)**

* Router translates many private IPs ↔ one public IP
* That's why your entire home/office shares one public IP
* Interview line: "NAT conserves IPv4 addresses and hides internal devices from the internet"

## 14. VLANs — Logical Separation (intro)

**Think:** office floors — Floor 1 HR = VLAN 10, Floor 2 Finance = VLAN 20, Floor 3 IT = VLAN 30

* One physical switch, multiple logical networks
* Benefits: security, organization, reduced broadcast traffic
* A VLAN = its own broadcast domain — devices in different VLANs need a router (inter-VLAN routing)

## 15. Troubleshooting Mindset

* Professional loop: **Observe → Investigate → Diagnose → Fix → Verify → Save**
* Never guess — collect evidence first; never configure before inspecting

**Basic tools**

| Tool | What it tells you |
| --- | --- |
| `ping` | is the host reachable |
| `traceroute` / `tracert` | where along the path it breaks |
| `ipconfig` / `ifconfig` | your IP, gateway, DNS settings |
| `show running-config` / `show vlan` / `show interfaces status` | switch state (Cisco) |

* Interview story: "First I check the current state with show/ping, change ONE thing, verify, then save the config."

**Scenario drills**

* Website won't open, but internet works? → DNS problem first — name→IP lookup is failing
* Two PCs on the same switch can't ping each other? → different VLANs/subnets, or a port is down
* Whole office offline? → think router/gateway/ISP, not individual devices
* What breaks if DHCP is down? → new devices get no IP — existing leases keep working for a while

**Red flags to avoid**

* Confusing MAC vs IP, switch vs router
* Can't narrate the YouTube journey
* No structured troubleshooting order (guessing)
